```python
## Demographics

### Gender Representation
labels = ['Women', 'Men', 'Non-binary/Other']
gender_data = [222, 170, 8]
plt.pie(gender_data, labels = labels, autopct='%1.1f%%')

# Women are pretty evenly represented in this study
```

```python
plt.figure(figsize=(12,4))
sns.set_theme(style='darkgrid')
plt.subplot(1,2,1)
sns.histplot(df['age'], color='purple', kde=True)
plt.title('Age')
plt.xlabel("Age")
plt.ylabel("Count of Participants")
plt.axvline(df['age'].mean(), color='Black', linestyle="dashed")

# Average age is around 30
# Could be outlier point around age 60
# We know they used minors in this study as well
```

```python
# How Well People Slept and for How Long

plt.scatter(df['sleep_hours'], df['sleep_quality_1_5'], color="Purple")
plt.title("How Well Participants Slept and for How Long")
plt.xlabel("Hours of Sleep")
plt.ylabel("Quality of Sleep Rating")

# It's clear that sleep quality improves when hours slept increases
# This may be good to tie into the story as screen time hours cut into sleeping hours
```

```python
# Screen Time 

plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
sns.histplot(df['screen_time_hours'], color='purple', kde=True)
plt.title('Screen Time (hours/day)')
plt.xlabel("Hours/Day")
plt.ylabel("Count of Participants")
plt.axvline(df['screen_time_hours'].mean(), color='Black', linestyle="dashed")

# The average screen time hours is higher than the average hours slept
```

```python
# Mental Wellness Index

plt.subplot(1,1,1)
sns.histplot(df['mental_wellness_index_0_100'], color='Purple', kde=True)
plt.axvline(df['mental_wellness_index_0_100'].mean(), color='Black', linestyle="dashed")
plt.title('Mental Wellness Index')
plt.xlabel("Mental Wellness Index (0 - 100)")
plt.ylabel("Count of Participants")

# Mental Wellness is hurting, with an average near 20 on a scale of 100
# It will be interesting to dig into which factors play the biggest part in moving that needle
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='mental_wellness_index_0_100', y='screen_time_hours', data=df)
plt.title('Screen time vs Mental Wellness')
plt.ylabel("Screen Time (Hours)")
plt.xlabel("Mental Wellness Index (1 - 100)")
sns.regplot(x='mental_wellness_index_0_100', y='screen_time_hours', data=df, color="Purple",)

# Regression line shows that mental wellness improves as screen time decreases
```

```python
# Screen Time by Occupation

plt.figure(figsize=(9,5))
sns.boxplot(x='occupation', y='screen_time_hours', data=df, color="Purple")
plt.title('Screen Time by Occupation')
plt.xlabel("Occupation")
plt.ylabel("Screen Time (Hours)")

# Students have the highest screen time. Is it simply because of their age or something else?
# Unemployed people have the lowest. Are they actively looking for work? If so, online or in person?
```
